```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.decomposition import NMF
```

```python
def generate_matrix(m, r, seed):
    """
    Generate a synthetic nonnegative matrix X = X1 @ X2
    """
    np.random.seed(seed)
    X1 = np.abs(np.random.randn(m, r))
    X2 = np.abs(np.random.randn(r, m))
    return X1 @ X2
```

```python
def project_NMF(V, r, solver, seed,
        max_iter=200, eps=1e-9,
        record_error=True):
    """
    Generic NMF framework using the provided solver (e.g., ALS).
    """
    rng = np.random.default_rng(seed)
    m, n = V.shape
    W = np.maximum(rng.random((m, r)), eps)
    H = np.maximum(rng.random((r, n)), eps)

    errors = []

    for t in range(max_iter):
        W, H = solver(V, W, H, eps=eps)
        if record_error:
            err = np.linalg.norm(V - (W@H), 'fro')**2
            errors.append(err)
    print(errors)
    if record_error:
        return W, H, errors
    else:
        return W, H
```

```python
def mult_update_step(V, W, H, eps=1e-9):
    """
    One multiplicative update iteration for NMF.
    Matches the structure of the ALS solver so it can be used inside NMF().
    """
    W = W*((V@np.transpose(H))/(W@H@np.transpose(H)+eps))
    H = H*((np.transpose(W)@V)/(np.transpose(W)@W@H+eps))

    return W, H
```

```python
seed_today=11122025

X = generate_matrix(100, 2, seed_today)
model = NMF(n_components=1, init='random', solver='mu', random_state=0,max_iter=100)
A = model.fit_transform(X)                                              #access the NMF factor matrices
S = model.components_
error = np.linalg.norm(X - (A@S),'fro')**2 # ||X - AS||_2
print(error)

errors = []
A, S, errors = project_NMF(X, 2, solver=mult_update_step, seed=seed_today, max_iter=100)
error = np.linalg.norm(X - (A@S),'fro')**2 # ||X - AS||_2
print(error)
```

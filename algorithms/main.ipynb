<p> Exploring algorithmic complexity for different Nonnegative Matrix Factorization (NMF). We explore Multiplicative Updates (MU), Alternating Least Squares (ALS), and Coordinate Descent (CD) Algorithms </p>

```python
import time

import numpy as np
import matplotlib.pyplot as plt
```

<h4>Part 1: Define Algorithms</h4>

```python
def generate_matrix(m, r, seed):
    """
    Generate a synthetic nonnegative matrix X = X1 @ X2
    Arguments:
    - m: dimension for the matrix
    - r: rank
    - seed: seed for np.random
    """
    np.random.seed(seed)
    X1 = np.abs(np.random.randn(m, r))
    X2 = np.abs(np.random.randn(r, m))
    return X1 @ X2

def project_NMF(V, r, solver, seed,
        max_iter=200,
        record_error=True):
    """
    Generic NMF framework using the provided solver (e.g., ALS).
    V: the matrix to be factored
    r: rank
    solver: NMF algorithm to use
    seed: seed for np.random
    max_iter: the number of steps to take (for gradient descent)
    record_error: bool to record the errors as we go
    """
    # to prevent divide by 0 errors
    eps=1e-9
    rng = np.random.default_rng(seed)
    m, n = V.shape

    # initialize W and H
    W = np.maximum(rng.random((m, r)), eps)
    H = np.maximum(rng.random((r, n)), eps)

    errors = []
    times = []
    num_iterations = -1

    start_time = time.time()

    for t in range(max_iter):
        W, H = solver(V, W, H)
        if record_error:
            err = np.linalg.norm(V - (W@H), 'fro')**2
            if t>0 and num_iterations == -1 and abs(err-errors[-1]) < 1e-4:
                num_iterations = t+1
            errors.append(err)

            elapsed = time.time() - start_time
            times.append(elapsed)
    if record_error:
        return W, H, errors, num_iterations, times
    else:
        return W, H
```

```python
def als_update_step(V, W, H):
    """
    A more stable ALS update:
    - Least squares updates
    - Projection to nonnegativity
    - Column normalization of W
    - Rescaling of H to match
    """
    # to prevent divide by 0 errors
    eps=1e-9
    r = W.shape[1]

    # === Update H ===
    WtW = W.T @ W + eps * np.eye(r)
    H = np.linalg.solve(WtW, W.T @ V)
    H = np.maximum(H, 0)

    # === Update W ===
    HHt = H @ H.T + eps * np.eye(r)
    W = (V @ H.T) @ np.linalg.inv(HHt)
    W = np.maximum(W, 0)

    # === Normalize W columns to avoid exploding scales ===
    norms = np.linalg.norm(W, axis=0)
    norms = np.maximum(norms, eps)   # avoid divide by zero
    W = W / norms
    H = H * norms[:, None]           # scale H to keep WH unchanged

    return W, H
```

```python
def mult_update_step(V, W, H):
    """
    One multiplicative update iteration for NMF.
    Matches the structure of the ALS solver so it can be used inside NMF().
    """
    # to prevent divide by 0 errors
    eps=1e-9

    W = W*((V@np.transpose(H))/(W@H@np.transpose(H)+eps))
    H = H*((np.transpose(W)@V)/(np.transpose(W)@W@H+eps))

    return W, H
```

```python
def cd_update_step(V, W, H):
    """
    One HALS (coordinate descent) update for NMF.
    Updates one column of W and one row of H at a time.
    """
    # to prevent divide by 0 errors
    eps=1e-9

    m, r = W.shape
    r, n = H.shape

    # === Update H row-by-row ===
    WH = W @ H
    for k in range(r):
        # residual excluding component k
        Hk = H[k, :] + (W[:, k].T @ (V - WH)) / (np.linalg.norm(W[:, k])**2 + eps)
        H[k, :] = np.maximum(Hk, 0)
        WH = W @ H   # update WH incrementally

    # === Update W column-by-column ===
    WH = W @ H
    for k in range(r):
        Wk = W[:, k] + ((V - WH) @ H[k, :].T) / (np.linalg.norm(H[k, :])**2 + eps)
        W[:, k] = np.maximum(Wk, 0)
        WH = W @ H

    return W, H
```

```python
# ===== CALL CELL =====

# Generate matrix using today's date as seed
seed_today = 11132025
X = generate_matrix(m=100, r=5, seed=seed_today)

# Same initialization for both algorithms
seed = seed_today // 2


def run_algs(M, seed_nmf, iters, rank):
    # Run stable ALS
    _, _, errs_als, _, times_als = project_NMF(V=M, r=rank, solver=als_update_step,
                        seed=seed_nmf, max_iter=iters)

    # Run multiplicative update
    _, _, errs_mu, _, times_mu = project_NMF(V=M, r=rank, solver=mult_update_step,
                        seed=seed_nmf, max_iter=iters)

    # Run coordinate descent
    _, _, errs_cd, _, times_cd = project_NMF(V=M, r=rank, solver=cd_update_step,
                        seed=seed_nmf, max_iter=iters)
    
    return errs_als, errs_mu, errs_cd, times_als, times_mu, times_cd


errs_als, errs_mu, errs_cd, _, _, _ = run_algs(M=X, seed_nmf = seed, iters=6000, rank=5)
# Print some diagnostic values
print("ALS_stable final error:", errs_als[-1])
print("MU final error:", errs_mu[-1])
print("CD final error:", errs_cd[-1])
print("ALS_stable first 5 errors:", errs_als[:5])
print("MU first 5 errors:", errs_mu[:5])
print("CD first 5 errors:", errs_cd[:5])
```

<h4>Part 2: Plot errors over iteration<h4>

```python
# run with rank 2
errs_als_2, errs_mu_2, errs_cd_2, _, _, _ = run_algs(M=X, seed_nmf = seed, iters=100, rank=2)

# run with rank 3
errs_als_3, errs_mu_3, errs_cd_3, _, _, _ = run_algs(M=X, seed_nmf = seed, iters=100, rank=3)

# run with rank 5
errs_als_5, errs_mu_5, errs_cd_5, _, _, _ = run_algs(M=X, seed_nmf = seed, iters=100, rank=5)
```

```python
def convergence_comparison(errs_als, errs_mu, errs_cd, rank):
    fig, axes = plt.subplots(1, 3, figsize=(16, 4), sharey=True)

    # ALS stable
    axes[0].plot(range(1, len(errs_als)+1), errs_als,
                marker='o', linewidth=1.8, label="ALS-stable")
    axes[0].set_title("ALS (stable)")
    axes[0].set_xlabel("Iteration")
    axes[0].set_ylabel("Reconstruction Error")
    axes[0].grid(True, linestyle='--', alpha=0.6)

    # MU
    axes[1].plot(range(1, len(errs_mu)+1), errs_mu,
                marker='o', linewidth=1.8, color='orange', label="MU")
    axes[1].set_title("Multiplicative Update")
    axes[1].set_xlabel("Iteration")
    axes[1].grid(True, linestyle='--', alpha=0.6)

    # CD
    axes[2].plot(range(1, len(errs_cd)+1), errs_cd,
                marker='o', linewidth=1.8, color='green', label="CD")
    axes[2].set_title("Coordinate Descent")
    axes[2].set_xlabel("Iteration")
    axes[2].grid(True, linestyle='--', alpha=0.6)

    plt.suptitle(f"NMF Convergence Comparison (Stable ALS vs MU vs CD) with rank {rank}", fontsize=14)
    plt.tight_layout()
    plt.show()

# plot with rank 2
convergence_comparison(errs_als_2, errs_mu_2, errs_cd_2, rank=2)
# plot with rank 3
convergence_comparison(errs_als_3, errs_mu_3, errs_cd_3, rank=3)
# plot with rank 5
convergence_comparison(errs_als_5, errs_mu_5, errs_cd_5, rank=5)
```

```python
fig, axes = plt.subplots(1, 3, figsize=(16, 4), sharey=True)

def log_convergence_comparison(errs_als, errs_mu, errs_cd, axes, rank):

    axes[0].plot(errs_als, marker='o', label=f"rank {rank}")
    axes[0].set_yscale('log')
    axes[0].set_title("ALS-stable (log-scale)")
    axes[0].set_xlabel("Iteration")
    axes[0].set_ylabel("Reconstruction Error")
    axes[0].grid(True, linestyle='--', alpha=0.6)
    axes[0].legend()

    axes[1].plot(errs_mu, marker='o', label=f"rank {rank}")
    axes[1].set_yscale('log')
    axes[1].set_title("MU (log-scale)")
    axes[1].set_xlabel("Iteration")
    axes[1].set_ylabel("Reconstruction Error")
    axes[1].grid(True, linestyle='--', alpha=0.6)
    axes[1].legend()

    axes[2].plot(errs_cd, marker='o', label=f"rank {rank}")
    axes[2].set_yscale('log')
    axes[2].set_title("CD (log-scale)")
    axes[2].set_xlabel("Iteration")
    axes[2].set_ylabel("Reconstruction Error")
    axes[2].grid(True, linestyle='--', alpha=0.6)
    axes[2].legend()

# plot with rank 2
log_convergence_comparison(errs_als_2, errs_mu_2, errs_cd_2, rank=2, axes=axes)
# plot with rank 3
log_convergence_comparison(errs_als_3, errs_mu_3, errs_cd_3, rank=3, axes=axes)
# plot with rank 5
log_convergence_comparison(errs_als_5, errs_mu_5, errs_cd_5, rank=5, axes=axes)

plt.suptitle(f"NMF Convergence (Log Scale, Same Initialization)")
plt.tight_layout()
plt.legend()
plt.show()
```

```python
fig, axes = plt.subplots(1, 3, figsize=(16, 4), sharey=True)

def no_first_iteration(errs_als, errs_mu, errs_cd, axes, rank):

    axes[0].plot(errs_als[1:], marker='o', label=f"rank {rank}")
    axes[0].set_yscale('log')
    axes[0].set_title("ALS-stable (log-scale)")
    axes[0].set_xlabel("Iteration")
    axes[0].set_ylabel("Reconstruction Error")
    axes[0].grid(True, linestyle='--', alpha=0.6)
    axes[0].legend()

    axes[1].plot(errs_mu[1:], marker='o', label=f"rank {rank}")
    axes[1].set_yscale('log')
    axes[1].set_title("MU (log-scale)")
    axes[1].set_xlabel("Iteration")
    axes[1].set_ylabel("Reconstruction Error")
    axes[1].grid(True, linestyle='--', alpha=0.6)
    axes[1].legend()

    axes[2].plot(errs_cd[1:], marker='o', label=f"rank {rank}")
    axes[2].set_yscale('log')
    axes[2].set_title("CD (log-scale)")
    axes[2].set_xlabel("Iteration")
    axes[2].set_ylabel("Reconstruction Error")
    axes[2].grid(True, linestyle='--', alpha=0.6)
    axes[2].legend()

# plot with rank 2
no_first_iteration(errs_als_2, errs_mu_2, errs_cd_2, rank=2, axes=axes)
# plot with rank 3
no_first_iteration(errs_als_3, errs_mu_3, errs_cd_3, rank=3, axes=axes)
# plot with rank 5
no_first_iteration(errs_als_5, errs_mu_5, errs_cd_5, rank=5, axes=axes)

plt.suptitle(f"NMF Convergence Comparison (Log Scale, Excluding First Iteration)")
plt.tight_layout()
plt.legend()
plt.show()
```

```python
fig, axes = plt.subplots(1, 3, figsize=(16, 4), sharey=True)

def compare_algs(errs_als, errs_mu, errs_cd, ax, rank):
    # Plot all three
    ax.plot(errs_als[1:], label="ALS")
    ax.plot(errs_mu[1:], label="MU")
    ax.plot(errs_cd[1:], label="CD")
    ax.set_xlabel("Iteration")
    ax.set_ylabel("Error")
    ax.set_title(f"Rank {rank}")
    ax.grid(True, linestyle="--", alpha=0.6)
    ax.legend()

compare_algs(errs_als_2, errs_mu_2, errs_cd_2, axes[0], rank=2)
compare_algs(errs_als_3, errs_mu_3, errs_cd_3, axes[1], rank=3)
compare_algs(errs_als_5, errs_mu_5, errs_cd_5, axes[2], rank=5)

plt.suptitle(f"NMF Convergence Comparison")
plt.tight_layout()
plt.legend()
plt.show()
```

<h4> Part 3: Wallclock timing of the algorithms </h4>

```python
def clock_time(r, solver, seed, iters):
    V = generate_matrix(m=100, r=r, seed=seed)
    times = []
    
    for i in range(iters):
        st_time =  time.time_ns()
        project_NMF(V, r, solver, seed, 50)
        end_time = time.time_ns()
        times.append(end_time-st_time)
    return times

fig, axes = plt.subplots(1, 3, figsize=(16, 4), sharey=True)
def compare_times(times_als, times_mu, times_cd, ax, rank):
    ax.plot(times_als, color="blue", label="ALS")
    ax.plot(times_mu, color="orange", label="MU")
    ax.plot(times_cd, color="green", label="CD")
    ax.set_xlabel("Iteration")
    ax.set_ylabel("Wallclock Time")
    ax.set_title(f"Rank {rank}")
    ax.grid(True, linestyle="--", alpha=0.6)
    ax.legend()


iterations = 200

times_als_2 = np.array(clock_time(r=2, solver=als_update_step, seed=seed_today, iters=iterations))
times_als_3= np.array(clock_time(r=3, solver=als_update_step, seed=seed_today, iters=iterations))
times_als_5 = np.array(clock_time(r=5, solver=als_update_step, seed=seed_today, iters=iterations))

times_mu_2 = np.array(clock_time(r=2, solver=mult_update_step, seed=seed_today, iters=iterations))
times_mu_3 = np.array(clock_time(r=3, solver=mult_update_step, seed=seed_today, iters=iterations))
times_mu_5 = np.array(clock_time(r=5, solver=mult_update_step, seed=seed_today, iters=iterations))

times_cd_2 = np.array(clock_time(r=2, solver=cd_update_step, seed=seed_today, iters=iterations))
times_cd_3 = np.array(clock_time(r=3, solver=cd_update_step, seed=seed_today, iters=iterations))
times_cd_5 = np.array(clock_time(r=5, solver=cd_update_step, seed=seed_today, iters=iterations))

compare_times(times_als_2, times_mu_2, times_cd_2, rank=2, ax=axes[0])
compare_times(times_als_3, times_mu_3, times_cd_3, rank=3, ax=axes[1])
compare_times(times_als_5, times_mu_2, times_cd_2, rank=5, ax=axes[2])
    
plt.suptitle(f"Wallclock Time Comparison")
plt.tight_layout()
plt.legend()
plt.show()


    
```

<h4>Part 4: Rank versus Iterations to Convergence</h4>

```python
ranks = [2, 3, 4, 5]

iters_als = []
iters_mu = []
iters_cd = []

seed_today = 11132025

for r in ranks:

    X_r = generate_matrix(m=100, r=r, seed=seed_today)
    seed = seed_today // 2

    _, _, _, iterations_als, _ = project_NMF(
        V=X_r, r=r, solver=als_update_step,
        seed=seed, max_iter=6000
    )
    iters_als.append(iterations_als)

    _, _, _, iterations_mu_alg, _ = project_NMF(
        V=X_r, r=r, solver=mult_update_step,
        seed=seed, max_iter=6000
    )
    iters_mu.append(iterations_mu_alg)

    _, _, _, iterations_cd_alg, _ = project_NMF(
        V=X_r, r=r, solver=cd_update_step,
        seed=seed, max_iter=6000
    )
    iters_cd.append(iterations_cd_alg)

def compare_iterations(ax, ranks, iters, title, color):
    ax.plot(ranks, iters, marker='o', color=color)
    ax.set_xlabel("Rank")
    ax.set_ylabel("Iterations to Convergence")
    ax.set_title(title)
    ax.set_xticks(ranks)
    ax.grid(True, linestyle="--", alpha=0.6)

fig, axes = plt.subplots(1, 3, figsize=(16, 4), sharey=True)

compare_iterations(axes[0], ranks, iters_als, "ALS", "blue")
compare_iterations(axes[1], ranks, iters_mu,  "MU",  "orange")
compare_iterations(axes[2], ranks, iters_cd,  "CD",  "green")

fig.suptitle("Iterations to Convergence vs Rank", fontsize=14)
plt.tight_layout()
plt.show()

```

<h4>Part 5: Iterations versus Reconstruction Error (model rank = true data rank)</h4>

```python
# run with rank 2
X_2 = generate_matrix(m=100, r=2, seed=seed_today)
errs_als_2, errs_mu_2, errs_cd_2, times_als_2, times_mu_2, times_cd_2 = run_algs(M=X_2, seed_nmf = seed, iters=100, rank=2)

# run with rank 3
X_3 = generate_matrix(m=100, r=3, seed=seed_today)
errs_als_3, errs_mu_3, errs_cd_3, times_als_3, times_mu_3, times_cd_3 = run_algs(M=X_3, seed_nmf = seed, iters=100, rank=3)

# run with rank 5
X_5 = generate_matrix(m=100, r=5, seed=seed_today)
errs_als_5, errs_mu_5, errs_cd_5, times_als_5, times_mu_5, times_cd_5 = run_algs(M=X_5, seed_nmf = seed, iters=100, rank=5)

fig, axes = plt.subplots(1, 3, figsize=(16, 4), sharey=True)

# plot with rank 2
no_first_iteration(errs_als_2, errs_mu_2, errs_cd_2, rank=2, axes=axes)
# plot with rank 3
no_first_iteration(errs_als_3, errs_mu_3, errs_cd_3, rank=3, axes=axes)
# plot with rank 5
no_first_iteration(errs_als_5, errs_mu_5, errs_cd_5, rank=5, axes=axes)

plt.suptitle(f"NMF Convergence Comparison with Model Rank = Data Rank (Log Scale, Excluding First Iteration)")
plt.tight_layout()
plt.legend()
plt.show()
```

<h4>Part 6: Wallclock time versus Reconstruction Error</h4>

```python
def plot_err_vs_time(ax, times_als, errors_als,
                          times_mu,  errors_mu,
                          times_cd,  errors_cd,
                          rank):
    ax.plot(times_als[1:], errors_als[1:], color="blue",   label="ALS")
    ax.plot(times_mu[1:],  errors_mu[1:],  color="orange", label="MU")
    ax.plot(times_cd[1:],  errors_cd[1:],  color="green",  label="CD")
    ax.set_xlabel("Wallclock Time")
    ax.set_ylabel("Reconstruction Error")
    ax.set_title(f"Rank {rank}")
    ax.grid(True, linestyle="--", alpha=0.6)
    ax.legend()


fig, axes = plt.subplots(1, 3, figsize=(18, 5), sharey=True)

# Rank 2 plot
plot_err_vs_time(
    axes[0],
    times_als_2, errs_als_2,
    times_mu_2, errs_mu_2,
    times_cd_2, errs_cd_2,
    2
)

# Rank 3 plot
plot_err_vs_time(
    axes[1],
    times_als_3, errs_als_3,
    times_mu_3, errs_mu_3,
    times_cd_3, errs_cd_3,
    3
)

# Rank 5 plot
plot_err_vs_time(
    axes[2],
    times_als_5, errs_als_5,
    times_mu_5, errs_mu_5,
    times_cd_5, errs_cd_5,
    5
)

fig.suptitle("Error vs Wallclock Time", fontsize=16)
plt.tight_layout()
plt.show()
```
